```python
# ============================================================
# PIMA DIABETES RESEARCH IMPLEMENTATION
# PAPER-INSPIRED ENSEMBLE APPROACH
# 5-FOLD STRATIFIED CV + SMOTETOMEK + 6 ML MODELS
# + WEIGHTED SOFT VOTING (RF + XGBOOST)
# ============================================================


# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip install -q imbalanced-learn xgboost


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from collections import Counter

from sklearn.model_selection import (
    StratifiedKFold,
    train_test_split
)

from sklearn.preprocessing import MinMaxScaler

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier
)

from sklearn.svm import SVC

from xgboost import XGBClassifier

from imblearn.combine import SMOTETomek

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


# Reproducibility
RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
tf.keras.utils.set_random_seed(RANDOM_STATE)


# ============================================================
# 3. LOAD DATASET
# ============================================================

from google.colab import files

uploaded = files.upload()

# Define the standard column names for the Pima Indians Diabetes dataset
column_names = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

# Load the dataset. If the first row contains numeric values (no header),
# we read it with header=None and manually set the names.
test_df = pd.read_csv("/content/pima-indians-diabetes.csv", nrows=1)
try:
    # Try to see if the first row elements are numbers, which indicates missing headers
    float(test_df.columns[0])
    df = pd.read_csv("/content/pima-indians-diabetes.csv", header=None, names=column_names)
except ValueError:
    # Otherwise, load normally
    df = pd.read_csv("/content/pima-indians-diabetes.csv")

print("\n========== DATASET LOADED ==========")

print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 4. EXPLORATORY DATA ANALYSIS
# ============================================================

print("\n========== DATASET INFORMATION ==========")

df.info()

print("\nStatistical Summary:")

display(df.describe())

print("\nMissing Values:")

print(df.isnull().sum())

print("\nOutcome Distribution:")

print(df["Outcome"].value_counts())


plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Outcome")

plt.title("Diabetes Outcome Distribution")

plt.show()


# ============================================================
# 5. HANDLE INVALID ZERO VALUES
# ============================================================

invalid_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for col in invalid_zero_columns:

    df[col] = df[col].replace(0, np.nan)


print("\n========== MISSING VALUES AFTER ZERO HANDLING ==========")

print(df[invalid_zero_columns].isnull().sum())


# ============================================================
# 6. CLASS-SPECIFIC MEDIAN IMPUTATION
# ============================================================

for col in invalid_zero_columns:

    for outcome_class in [0, 1]:

        median_value = df.loc[
            df["Outcome"] == outcome_class,
            col
        ].median()

        df.loc[
            (df["Outcome"] == outcome_class) &
            (df[col].isnull()),
            col
        ] = median_value


print("\n========== AFTER MEDIAN IMPUTATION ==========")

print(df.isnull().sum())


# ============================================================
# 7. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop("Outcome", axis=1)

y = df["Outcome"]

print("\nFeature shape:", X.shape)

print("Target shape:", y.shape)

print("\nTarget distribution:")

print(y.value_counts())


# ============================================================
# 8. CREATE 5-FOLD STRATIFIED CROSS VALIDATION
# ============================================================

skf = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=RANDOM_STATE

)


print("\n========== FIVE FOLD DISTRIBUTION ==========")

for fold, (train_index, test_index) in enumerate(
    skf.split(X, y), 1
):

    print(f"\nFOLD {fold}")

    print(
        "Train:",
        y.iloc[train_index].value_counts().to_dict()
    )

    print(
        "Test:",
        y.iloc[test_index].value_counts().to_dict()
    )


# ============================================================
# 9. TRAIN SIX CLASSIFICATION MODELS
#    APPLY SMOTETOMEK ONLY TO TRAINING FOLD
# ============================================================

results = []

fold_data = []


for fold, (train_index, test_index) in enumerate(
    skf.split(X, y), 1
):

    print(f"\n\n========== MODEL TRAINING FOLD {fold} ==========")


    # Split training and testing data

    X_train = X.iloc[train_index].copy()

    X_test = X.iloc[test_index].copy()

    y_train = y.iloc[train_index].copy()

    y_test = y.iloc[test_index].copy()


    # SMOTETomek

    print("Before SMOTETomek:", Counter(y_train))

    smt = SMOTETomek(
        random_state=RANDOM_STATE
    )

    X_train_resampled, y_train_resampled = (
        smt.fit_resample(
            X_train,
            y_train
        )
    )

    print(
        "After SMOTETomek:",
        Counter(y_train_resampled)
    )


    # Store fold data for later use

    fold_data.append({

        "fold": fold,

        "X_train": X_train_resampled,

        "y_train": y_train_resampled,

        "X_test": X_test,

        "y_test": y_test

    })


    # MinMax Scaling

    scaler = MinMaxScaler()

    X_train_scaled = scaler.fit_transform(
        X_train_resampled
    )

    X_test_scaled = scaler.transform(
        X_test
    )


    # Define five ML models

    models = {

        "Logistic Regression": LogisticRegression(
            C=10,
            max_iter=1000,
            random_state=RANDOM_STATE
        ),

        "Random Forest": RandomForestClassifier(
            random_state=RANDOM_STATE
        ),

        "AdaBoost": AdaBoostClassifier(
            n_estimators=300,
            learning_rate=0.01,
            random_state=RANDOM_STATE
        ),

        "XGBoost": XGBClassifier(
            n_estimators=20,
            random_state=RANDOM_STATE,
            eval_metric="logloss"
        ),

        "SVM": SVC()

    }


    # Train and evaluate five ML models

    for model_name, model in models.items():

        print("\nTraining:", model_name)


        # LR and SVM use scaled features

        if model_name in [
            "Logistic Regression",
            "SVM"
        ]:

            model.fit(
                X_train_scaled,
                y_train_resampled
            )

            y_pred = model.predict(
                X_test_scaled
            )

            if hasattr(model, "predict_proba"):

                y_score = model.predict_proba(
                    X_test_scaled
                )[:, 1]

            else:

                y_score = model.decision_function(
                    X_test_scaled
                )


        # Tree models use unscaled features

        else:

            model.fit(
                X_train_resampled,
                y_train_resampled
            )

            y_pred = model.predict(
                X_test
            )

            y_score = model.predict_proba(
                X_test
            )[:, 1]


        # Calculate metrics

        accuracy = accuracy_score(
            y_test,
            y_pred
        )

        precision = precision_score(
            y_test,
            y_pred,
            zero_division=0
        )

        recall = recall_score(
            y_test,
            y_pred,
            zero_division=0
        )

        f1 = f1_score(
            y_test,
            y_pred,
            zero_division=0
        )

        auc = roc_auc_score(
            y_test,
            y_score
        )


        # Save results

        results.append({

            "Fold": fold,

            "Model": model_name,

            "Accuracy": accuracy,

            "Precision": precision,

            "Recall": recall,

            "F1 Score": f1,

            "ROC AUC": auc

        })


        print(
            "Accuracy:", round(accuracy, 4),
            "| F1:", round(f1, 4)
        )


    # ========================================================
    # 10. ARTIFICIAL NEURAL NETWORK
    # ========================================================

    print("\nTraining ANN...")


    ann = tf.keras.Sequential([

        tf.keras.layers.Input(
            shape=(X_train_scaled.shape[1],)
        ),

        tf.keras.layers.Dense(
            10,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            8,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            1,
            activation="sigmoid"
        )

    ])


    ann.compile(

        optimizer=tf.keras.optimizers.Adam(),

        loss="binary_crossentropy",

        metrics=["accuracy"]

    )


    ann.fit(

        X_train_scaled,

        y_train_resampled,

        epochs=20,

        batch_size=5,

        verbose=0

    )


    ann_prob = ann.predict(
        X_test_scaled,
        verbose=0
    ).ravel()


    ann_pred = (
        ann_prob >= 0.5
    ).astype(int)


    results.append({

        "Fold": fold,

        "Model": "ANN",

        "Accuracy": accuracy_score(
            y_test,
            ann_pred
        ),

        "Precision": precision_score(
            y_test,
            ann_pred,
            zero_division=0
        ),

        "Recall": recall_score(
            y_test,
            ann_pred,
            zero_division=0
        ),

        "F1 Score": f1_score(
            y_test,
            ann_pred,
            zero_division=0
        ),

        "ROC AUC": roc_auc_score(
            y_test,
            ann_prob
        )

    })


    print("ANN completed!")


# ============================================================
# 11. INDIVIDUAL MODEL RESULTS
# ============================================================

results_df = pd.DataFrame(results)

print("\n========== ALL MODEL RESULTS ==========")

display(results_df.round(4))


# ============================================================
# 12. FIVE-FOLD PERFORMANCE SUMMARY
# ============================================================

summary = results_df.groupby("Model").agg({

    "Accuracy": ["mean", "std"],

    "Precision": ["mean", "std"],

    "Recall": ["mean", "std"],

    "F1 Score": ["mean", "std"],

    "ROC AUC": ["mean", "std"]

})

print("\n========== INDIVIDUAL MODEL SUMMARY ==========")

display(summary.round(4))


# ============================================================
# 13. VISUALIZE INDIVIDUAL MODEL ACCURACY
# ============================================================

mean_results = (

    results_df.groupby("Model")["Accuracy"]

    .mean()

    .sort_values(ascending=False)

)


plt.figure(figsize=(11, 6))

sns.barplot(
    x=mean_results.index,
    y=mean_results.values
)

plt.title("Average Accuracy Across 5 Folds")

plt.xlabel("Classification Models")

plt.ylabel("Mean Accuracy")

plt.xticks(rotation=30)

plt.ylim(0, 1)

plt.show()


# ============================================================
# 14. WEIGHTED SOFT VOTING
#     RANDOM FOREST + XGBOOST
# ============================================================

ensemble_results = []

selected_weights = []


for fold, (train_index, test_index) in enumerate(
    skf.split(X, y), 1
):

    print(f"\n\n========== ENSEMBLE FOLD {fold} ==========")


    # --------------------------------------------------------
    # OUTER TRAIN-TEST SPLIT
    # --------------------------------------------------------

    X_train = X.iloc[train_index].copy()

    X_test = X.iloc[test_index].copy()

    y_train = y.iloc[train_index].copy()

    y_test = y.iloc[test_index].copy()


    # --------------------------------------------------------
    # INNER TRAIN-VALIDATION SPLIT
    # --------------------------------------------------------

    X_inner_train, X_val, y_inner_train, y_val = (
        train_test_split(

            X_train,

            y_train,

            test_size=0.20,

            stratify=y_train,

            random_state=RANDOM_STATE

        )
    )


    # --------------------------------------------------------
    # SMOTETOMEK ON INNER TRAINING DATA
    # --------------------------------------------------------

    smt_inner = SMOTETomek(
        random_state=RANDOM_STATE
    )

    X_inner_resampled, y_inner_resampled = (
        smt_inner.fit_resample(
            X_inner_train,
            y_inner_train
        )
    )


    # --------------------------------------------------------
    # TRAIN RANDOM FOREST ON INNER TRAINING DATA
    # --------------------------------------------------------

    rf_inner = RandomForestClassifier(
        random_state=RANDOM_STATE
    )

    rf_inner.fit(
        X_inner_resampled,
        y_inner_resampled
    )


    # --------------------------------------------------------
    # TRAIN XGBOOST ON INNER TRAINING DATA
    # --------------------------------------------------------

    xgb_inner = XGBClassifier(

        n_estimators=20,

        random_state=RANDOM_STATE,

        eval_metric="logloss"

    )

    xgb_inner.fit(
        X_inner_resampled,
        y_inner_resampled
    )


    # --------------------------------------------------------
    # VALIDATION PROBABILITIES
    # --------------------------------------------------------

    rf_val_prob = rf_inner.predict_proba(
        X_val
    )[:, 1]

    xgb_val_prob = xgb_inner.predict_proba(
        X_val
    )[:, 1]


    # --------------------------------------------------------
    # SEARCH BEST WEIGHTS
    # --------------------------------------------------------

    best_accuracy = -1

    best_rf_weight = None

    best_xgb_weight = None


    for rf_weight in np.arange(
        0, 1.01, 0.1
    ):

        xgb_weight = 1 - rf_weight


        ensemble_val_prob = (

            rf_weight * rf_val_prob

            +

            xgb_weight * xgb_val_prob

        )


        ensemble_val_pred = (
            ensemble_val_prob >= 0.5
        ).astype(int)


        accuracy = accuracy_score(
            y_val,
            ensemble_val_pred
        )


        if accuracy > best_accuracy:

            best_accuracy = accuracy

            best_rf_weight = rf_weight

            best_xgb_weight = xgb_weight


    print("Best RF weight:", best_rf_weight)

    print("Best XGBoost weight:", best_xgb_weight)

    print(
        "Validation accuracy:",
        round(best_accuracy, 4)
    )


    # --------------------------------------------------------
    # RETRAIN USING FULL OUTER TRAINING FOLD
    # --------------------------------------------------------

    smt_final = SMOTETomek(
        random_state=RANDOM_STATE
    )

    X_train_resampled, y_train_resampled = (
        smt_final.fit_resample(
            X_train,
            y_train
        )
    )


    rf_final = RandomForestClassifier(
        random_state=RANDOM_STATE
    )

    xgb_final = XGBClassifier(

        n_estimators=20,

        random_state=RANDOM_STATE,

        eval_metric="logloss"

    )


    rf_final.fit(
        X_train_resampled,
        y_train_resampled
    )

    xgb_final.fit(
        X_train_resampled,
        y_train_resampled
    )


    # --------------------------------------------------------
    # OUTER TEST PROBABILITIES
    # --------------------------------------------------------

    rf_test_prob = rf_final.predict_proba(
        X_test
    )[:, 1]

    xgb_test_prob = xgb_final.predict_proba(
        X_test
    )[:, 1]


    # --------------------------------------------------------
    # WEIGHTED SOFT VOTING
    # --------------------------------------------------------

    ensemble_test_prob = (

        best_rf_weight * rf_test_prob

        +

        best_xgb_weight * xgb_test_prob

    )


    ensemble_test_pred = (
        ensemble_test_prob >= 0.5
    ).astype(int)


    # --------------------------------------------------------
    # ENSEMBLE EVALUATION
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        ensemble_test_pred
    )

    precision = precision_score(
        y_test,
        ensemble_test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        ensemble_test_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        ensemble_test_pred,
        zero_division=0
    )

    auc = roc_auc_score(
        y_test,
        ensemble_test_prob
    )


    ensemble_results.append({

        "Fold": fold,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1,

        "ROC AUC": auc

    })


    selected_weights.append({

        "Fold": fold,

        "RF Weight": best_rf_weight,

        "XGB Weight": best_xgb_weight

    })


    print(
        "Test Accuracy:",
        round(accuracy, 4)
    )

    print(
        "Test F1:",
        round(f1, 4)
    )


# ============================================================
# 15. ENSEMBLE RESULTS
# ============================================================

ensemble_df = pd.DataFrame(
    ensemble_results
)

weights_df = pd.DataFrame(
    selected_weights
)


print("\n========== ENSEMBLE FOLD RESULTS ==========")

display(ensemble_df.round(4))


print("\n========== SELECTED WEIGHTS ==========")

display(weights_df.round(4))


print("\n========== ENSEMBLE AVERAGE PERFORMANCE ==========")

display(

    ensemble_df.drop(
        columns=["Fold"]
    ).agg(["mean", "std"]).round(4)

)


# ============================================================
# 16. FINAL COMPARISON
# ============================================================

individual_summary = (

    results_df.groupby("Model")[

        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score",
            "ROC AUC"
        ]

    ].mean()

)


ensemble_summary = (

    ensemble_df[

        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score",
            "ROC AUC"
        ]

    ].mean()

)


comparison_df = individual_summary.copy()

comparison_df.loc[
    "Weighted Soft Voting"
] = ensemble_summary


print("\n========== FINAL MODEL COMPARISON ==========")

display(
    comparison_df.round(4)
)


# ============================================================
# 17. FINAL ACCURACY VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 6))

comparison_df["Accuracy"].sort_values(
    ascending=False
).plot(
    kind="bar"
)

plt.title(
    "Individual Models vs Weighted Soft Voting"
)

plt.xlabel("Classification Models")

plt.ylabel("Mean Accuracy")

plt.xticks(rotation=35)

plt.ylim(0, 1)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ============================================================
# 18. FINAL F1 SCORE VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 6))

comparison_df["F1 Score"].sort_values(
    ascending=False
).plot(
    kind="bar"
)

plt.title(
    "F1 Score Comparison Across Models"
)

plt.xlabel("Classification Models")

plt.ylabel("Mean F1 Score")

plt.xticks(rotation=35)

plt.ylim(0, 1)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ============================================================
# 19. SAVE RESULTS
# ============================================================

results_df.to_csv(
    "individual_model_results.csv",
    index=False
)

ensemble_df.to_csv(
    "ensemble_results.csv",
    index=False
)

weights_df.to_csv(
    "selected_ensemble_weights.csv",
    index=False
)

comparison_df.to_csv(
    "final_model_comparison.csv"
)


print("\n========================================")

print("RESEARCH IMPLEMENTATION COMPLETED!")

print("All result files have been saved.")

print("========================================")
```
